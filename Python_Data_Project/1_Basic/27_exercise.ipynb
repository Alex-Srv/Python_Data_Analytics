```python
df.sort_values(by='job_posted_date', inplace=True)

df.plot(x='job_posted_date', y='salary_year_avg', kind='line')
```

```python
df.sort_values('job_posted_date').plot(x='job_posted_date', y='salary_year_avg', kind='line')
```

```python
df['job_month'] = df['job_posted_date'].dt.to_period('M')

# 3. Группируем, например, для подсчета количества вакансий по месяцам
monthly_counts = df.groupby('job_month').size() # или агрегация по зарплате
print(monthly_counts)
```

```python
# 1. Группируем по месяцам и считаем среднюю зарплату
df_monthly = df.groupby('job_month')['salary_year_avg'].mean().reset_index()

# 2. Строим график (теперь точек всего по количеству месяцев)
df_monthly.plot(x='job_month', y='salary_year_avg', kind='bar')
```
