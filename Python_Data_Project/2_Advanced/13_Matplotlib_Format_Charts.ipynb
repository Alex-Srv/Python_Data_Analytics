```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
fix, ax = plt.subplots(2, 3)

df['job_title_short'].value_counts().plot(kind= 'bar', ax= ax[0 ,1])

df['job_schedule_type'].value_counts().head(3).plot(kind = 'bar', ax= ax[0, 0])
plt.tight_layout()
```

```python
df_skills = df.copy()
df_skills = df_skills.explode('job_skills')
skills_count = df_skills.groupby(['job_skills', 'job_title_short']).size()
df_skills_count = skills_count.reset_index(name='skills_count')
df_skills_count.sort_values(by='skills_count', ascending=False, inplace=True)
df_skills_count
```

```python
job_titles = ['Data Scientist', 'Data Analyst', 'Data Engineer']
```

```python
fig ,ax = plt.subplots(3,1)

for i , job_title in enumerate(job_titles):
    df_plot = df_skills_count[df_skills_count['job_title_short'] == job_title].head(5)
    df_plot.plot(kind='barh',x='job_skills', y= 'skills_count', ax=ax[i], title=job_title)
    ax[i].invert_yaxis()
    ax[i].set_ylabel('')
    ax[i].legend().set_visible(False)
    ax[i].set_xlim(0,120000)
    fig.suptitle('Counts of Tops Skills in Job Postings', fontsize=15)
    fig.tight_layout()
```
