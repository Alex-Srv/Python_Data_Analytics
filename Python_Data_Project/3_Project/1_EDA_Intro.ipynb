```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt
import seaborn as sns

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
```

```python
df_DA_RUS = df[(df['job_country'] == 'Russia') & (df['job_title_short'] == 'Data Analyst')]
```

```python
df_plot = df_DA_RUS['job_location'].value_counts().head(10).to_frame()
```

```python
sns.set_theme(style= 'ticks')

sns.barplot(data=df_plot, x = 'count', y = 'job_location', hue = 'count', palette= 'dark:g_r')

plt.legend('')
plt.title('Count of job locations in Russia per place')
plt.xlabel('Number of Jobs')
plt.gca().set_ylabel('')
```

```python
# rewrite the above with a for loop
dict_column = {
    'job_work_from_home': 'Work from Home Offered',
    'job_no_degree_mention': 'Degree Requirement',
    'job_health_insurance': 'Health Insurance Offered'
}

fig, ax = plt.subplots(1, 3)
fig.set_size_inches(12, 5)

for i, (column, title) in enumerate(dict_column.items()):
    counts = df_DA_RUS[column].value_counts()
    ax[i].pie(counts, labels = counts.index ,autopct='%1.1f%%')
    ax[i].set_title(title)
# plt.suptitle('Benefit Analysis of Data Jobs', fontsize=16)
plt.show()
```
